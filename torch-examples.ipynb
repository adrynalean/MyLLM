```python
#torch.multinomial

# define probability tensor
probabilities = torch.tensor([0.1,0.9]) # it assignes the index values to the tensor depending on the probability listed
# so in the tensor output there is a 10% change to get 0 (index ot 0.1) and 90% change of getting 1 (index of 90%)
samples = torch.multinomial(probabilities, num_samples=10, replacement=True)
samples
```

```python
#torch.tril

# trill stands for trangle lower

out = torch.tril(torch.ones(5,5))
out
```

```python
output = torch.triu(torch.ones(5,5))
output
```

```python
out = torch.zeros(5,5).masked_fill(torch.tril(torch.ones(5,5)) == 0, float('-inf')) # basically takes the number and find e^that number
# so e^0 becomes 1 e^-inf becomes 0 .... referred to as exponentiation
print("out - ", out)
out2 = torch.exp(out)
print("Out exp - ", out2)
```

```python
#torch.nn.linear

sample = torch.tensor([10.,10.,10.])
linear = nn.Linear(3,3,bias = False)
print(linear(sample))
```

```python
#torch.cat
# short for concatenate

tensor = torch.tensor([1,2,3,4])
out = torch.cat((tensor, torch.tensor([5])), dim=0)
out
```

```python
#torch.F.softmax
import torch.nn.functional as F
#create a tensor
tensor1 = torch.tensor([1. , 2. , 3.])

softmax_output = F.softmax(tensor1, dim=0)

print(softmax_output)
```

```python
# Embedding

# initialize embedding layer
vocab_size = 1000
embedding_dim = 100

embedding = nn.Embedding(vocab_size, embedding_dim)

# create input token indices
input_indices = torch.LongTensor([1, 5, 3, 2])

# apply embedding layer
embedded_output = embedding(input_indices)

# output shape
print(embedded_output.shape)
```
